```python
import numpy as np

import pandas as pd

from matplotlib import pyplot as plt

path = r"C:\Users\wongtik\Desktop\PCC\IPBR\Mystery_Gene.txt"

## define the path opened as geneCode
## replace all the T gene to U gene
with open(path, 'r') as geneCode:
    geneString = geneCode.read().replace('\n', '').replace('T', 'U')

# Group every 3 elements into a list of codons
# range parameters are: (start, stop, step)
codon_list = ["".join(geneString[i:i+3]) for i in range(0, len(geneString), 3)]

# Convert codons into a new numpy array
codonBases = np.array(codon_list)
```

```python
# Data of a codon dictionary. Each codon will make one animo acid. Shorthands for amino acid is used here
codon_dict = {
    # U
    'UUU': 'F', 'UUC': 'F', 'UUA': 'L', 'UUG': 'L',
    'UCU': 'S', 'UCC': 'S', 'UCA': 'S', 'UCG': 'S',
    'UAU': 'Y', 'UAC': 'Y', 'UAA': 'Stop', 'UAG': 'Stop',
    'UGU': 'C', 'UGC': 'C', 'UGA': 'Stop', 'UGG': 'W',
    # C
    'CUU': 'L', 'CUC': 'L', 'CUA': 'L', 'CUG': 'L',
    'CCU': 'P', 'CCC': 'P', 'CCA': 'P', 'CCG': 'P',
    'CAU': 'H', 'CAC': 'H', 'CAA': 'Q', 'CAG': 'Q',
    'CGU': 'R', 'CGC': 'R', 'CGA': 'R', 'CGG': 'R',
    # A
    'AUU': 'I', 'AUC': 'I', 'AUA': 'I', 'AUG': 'M',  # AUG is also the Start codon
    'ACU': 'T', 'ACC': 'T', 'ACA': 'T', 'ACG': 'T',
    'AAU': 'N', 'AAC': 'N', 'AAA': 'K', 'AAG': 'K',
    'AGU': 'S', 'AGC': 'S', 'AGA': 'R', 'AGG': 'R',
    # G
    'GUU': 'V', 'GUC': 'V', 'GUA': 'V', 'GUG': 'V',
    'GCU': 'A', 'GCC': 'A', 'GCA': 'A', 'GCG': 'A',
    'GAU': 'D', 'GAC': 'D', 'GAA': 'E', 'GAG': 'E',
    'GGU': 'G', 'GGC': 'G', 'GGA': 'G', 'GGG': 'G'
}
```

```python
# Translate every codon in the array using the dictionary
aminoAcids = [codon_dict[codon] for codon in codonBases if codon in codon_dict]

print("total number of animo acids:", len(aminoAcids))
print('\n')

print(aminoAcids[:len(aminoAcids)])
```

```python
# join the list of amino acids into 1 single string called aminoString
# Since the last codon is a STOP codon, -1 from the list to remove the STOP codon
aminoString = "".join(aminoAcids[:-1])

print (aminoString)
```

```python
#import Seq object class from the Biopython library
from Bio.Seq import Seq

#import online NCBIWWW and NCBLXML object class from the Bioblast library
from Bio.Blast import NCBIWWW
from Bio.Blast import NCBIXML
```

```python
#use the BLAST algorithm and stores the raw results in result_handle
result_handle = NCBIWWW.qblast("blastp", "nr", aminoString)
```

```python
#read the result_handle by NCBIXML
blast_record = NCBIXML.read(result_handle)
```

```python
#print the title of the result protein
blast_record.alignments[0].title
```

```python
#access the alignment details
for alignment in blast_record.alignments:
    for hsp in alignment.hsps:
        print(f'Sequence: {alignment.title}')
        print(f'length: {alignment.length}')
        print(f'e value: {hsp.expect}')
        print(f'Query: {hsp.query}')
        print(f'Subject: {hsp.sbjct}')
        print('\n')
```

```python
#close up the system resources
result_handle.close()
```
